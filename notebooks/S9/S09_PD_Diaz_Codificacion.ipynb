## Clasificación de mis columnas categóricas

| Columna | Tipo | Cardinalidad | Notas |
|---|---|---|---|
| `canal_compra` | Nominal | Baja (4) | Sin orden intrínseco entre canales |
| `metodo_pago` | Nominal | Baja (4) | Sin orden intrínseco entre métodos de pago |
| `ciudad_tienda` | Nominal | Baja (12) | Por debajo del umbral de 15; a diferencia del ejemplo genérico de la lección, en NovaMarket esta columna NO es de alta cardinalidad |
| `categoria_producto` | Nominal | Baja (6) | Sin orden intrínseco entre categorías |
| `producto` | Nominal | Alta (24) | Supera el umbral de 15; requiere técnica distinta a las columnas de baja cardinalidad |
| `nivel_satisfaccion` | Ordinal | Baja (3) | Tiene orden implícito (ej. bajo/medio/alto) |
| `nivel_lealtad` | Ordinal | Baja (4) | Tiene orden implícito (ej. bronce/plata/oro/platino) |
| `comentario_cliente` | Nominal | Baja (10) | Aunque el nombre sugiere texto libre, solo 10 valores únicos indica que son comentarios predefinidos/categóricos, no texto abierto |

**Excluidas del proceso de codificación categórica** (son identificadores o fechas, no categorías): `id_pedido`, `id_cliente`, `correo_cliente`, `fecha_pedido`, `fecha_actualizacion_stock`.

# 2. One hot encoding

```python
df_codificado = pd.get_dummies(df, columns=["categoria_producto", "metodo_pago", "canal_compra"],
                                 prefix=["cat", "pago", "canal"])
```

## One-hot encoding de variables nominales de baja cardinalidad

Se aplica `pd.get_dummies()` sobre `categoria_producto`, `metodo_pago`, `canal_compra` y , ya que las cuatro son variables nominales (sin orden intrínseco) con baja cardinalidad (6, 4, 4  categorías respectivamente, todas por debajo del umbral de 15). Se ajusta con `df` completo en vez de solo `df_entrenamiento`, dado que `OneHotEncoder`/`get_dummies` no aprende estadísticas continuas sensibles a fuga de datos (como media o mediana), sino únicamente las categorías existentes — el riesgo de fuga de datos aquí es mínimo comparado con el escalado numérico de la semana anterior.
